````python
# Example: Loading and exploring data structure
# Note: This assumes data files are in the data/ directory

try:
    from utils import load_data
    
    # Load all data
    data = load_data()
    
    print("📊 Available datasets:")
    for key, df in data.items():
        if not df.empty:
            print(f"  - {key}: {df.shape[0]} rows, {df.shape[1]} columns")
            print(f"    Columns: {list(df.columns)[:5]}...")
        else:
            print(f"  - {key}: Empty or not found")
            
except Exception as e:
    print(f"⚠️ Data loading example (requires data files): {e}")
    print("Please ensure CSV files are in the data/ directory")
````

````python
# Example: Creating a donut chart for medal distribution
# This demonstrates the visualization approach used in the Overview page

# Sample medal data
sample_medals = pd.DataFrame({
    'Medal': ['Gold', 'Silver', 'Bronze'],
    'Count': [300, 300, 300]
})

# Create donut chart
fig = px.pie(
    sample_medals,
    values='Count',
    names='Medal',
    hole=0.4,  # Creates donut chart
    title="Global Medal Distribution",
    color_discrete_map={'Gold': '#FFD700', 'Silver': '#C0C0C0', 'Bronze': '#CD7F32'}
)

fig.update_traces(textposition='inside', textinfo='percent+label')
fig.show()

print("✅ Donut chart example created!")
````

### 3. Page 2: Global Analysis (`pages/2_Global_Analysis.py`)

**Key Features:**
- World Medal Map (Choropleth)
- Medal Hierarchy (Sunburst & Treemap)
- Continent vs. Medals (Grouped Bar Chart)
- Top 20 Countries vs. Medals (Grouped Bar Chart)

**Implementation Highlights:**
- Continent mapping for geographical analysis
- Hierarchical data structure (Continent → Country → Sport)
- Interactive map visualizations


````python
# Example: Creating a hierarchical sunburst chart
# This demonstrates the approach used for medal hierarchy visualization

# Sample hierarchical data
hierarchy_data = pd.DataFrame({
    'Continent': ['Europe', 'Europe', 'Asia', 'Asia', 'North America'],
    'Country': ['France', 'Germany', 'China', 'Japan', 'USA'],
    'Sport': ['Swimming', 'Athletics', 'Swimming', 'Gymnastics', 'Basketball'],
    'Medal_Count': [10, 8, 15, 12, 9]
})

# Create sunburst chart
fig = px.sunburst(
    hierarchy_data,
    path=['Continent', 'Country', 'Sport'],
    values='Medal_Count',
    title="Medal Hierarchy by Continent",
    color='Medal_Count',
    color_continuous_scale='viridis'
)

fig.show()

print("✅ Sunburst chart example created!")
````

### 4. Page 3: Athlete Performance (`pages/3_Athlete_Performance.py`)

**Key Features:**
- Athlete Profile Card with search
- Age Distribution (Box & Violin Plots)
- Gender Distribution (World/Continent/Country)
- Top 10 Athletes by Medals

**Implementation Highlights:**
- Dynamic athlete search and selection
- Profile card with image placeholder
- Flexible grouping options (by sport or gender)


````python
# Example: Creating box and violin plots for age distribution
# This demonstrates the visualization approach used in the Athlete Performance page

# Sample athlete age data
np.random.seed(42)
athlete_data = pd.DataFrame({
    'Sport': np.random.choice(['Swimming', 'Athletics', 'Gymnastics'], 100),
    'Gender': np.random.choice(['M', 'F'], 100),
    'Age': np.random.normal(25, 5, 100).astype(int)
})

# Create box plot
fig_box = px.box(
    athlete_data,
    x='Sport',
    y='Age',
    title="Age Distribution by Sport",
    labels={'Age': 'Age (years)', 'Sport': 'Sport'}
)

fig_box.show()

# Create violin plot
fig_violin = px.violin(
    athlete_data,
    x='Sport',
    y='Age',
    title="Age Distribution (Violin) by Sport",
    box=True
)

fig_violin.show()

print("✅ Box and Violin plot examples created!")
````

### 5. Page 4: Sports and Events (`pages/4_Sports_and_Events.py`)

**Key Features:**
- Event Schedule (Timeline/Gantt Chart)
- Medal Count by Sport (Treemap)
- Venue Map (Scatter Mapbox)

**Implementation Highlights:**
- Timeline visualization for event scheduling
- Treemap for sport-based medal analysis
- Interactive map for venue locations


````python
# Example: Creating a timeline/Gantt chart for event schedule
# This demonstrates the approach used for event scheduling visualization

# Sample event schedule data
schedule_data = pd.DataFrame({
    'Event': ['100m Final', '200m Final', 'Swimming 100m', 'Basketball Final'],
    'Start': pd.to_datetime(['2024-07-28 10:00', '2024-07-28 14:00', 
                            '2024-07-29 09:00', '2024-07-30 20:00']),
    'End': pd.to_datetime(['2024-07-28 10:15', '2024-07-28 14:20',
                          '2024-07-29 09:30', '2024-07-30 22:00']),
    'Sport': ['Athletics', 'Athletics', 'Swimming', 'Basketball']
})

# Create timeline chart
fig = px.timeline(
    schedule_data,
    x_start='Start',
    x_end='End',
    y='Event',
    title="Event Schedule Timeline",
    color='Sport'
)

fig.update_yaxes(autorange="reversed")
fig.show()

print("✅ Timeline chart example created!")
````
